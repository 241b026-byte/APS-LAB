# Shortest Path Using Dijkstra Algorithm in Python
Problem Description:
Given a weighted graph with non negative edge weights. Find the shortest distance from a source vertex to all other vertices using Dijkstra algorithm

#Theory
Dijkstra algorithm finds shortest distance from one source node to all other nodes in a weighted graph with non negative weights. Dictionary structure stores nodes as keys and neighbor nodes with weights as values.

##Algorithm Steps

1. Represent graph using dictionary.
2. Set distance of every node as infinity.
3. Set distance of source node as 0.
4. Select node with smallest distance among unvisited nodes.
5. Update distance of neighbor nodes.
6. Repeat until all nodes get visited.

```python
def dijkstra(graph, start):

    distances = {}
    visited = []

    for node in graph:
        distances[node] = float('inf')

    distances[start] = 0

    while len(visited) < len(graph):

        min_node = None
        min_distance = float('inf')

        for node in graph:
            if node not in visited and distances[node] < min_distance:
                min_distance = distances[node]
                min_node = node

        visited.append(min_node)

        for neighbor, weight in graph[min_node].items():
            new_distance = distances[min_node] + weight

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance

    return distances
```

```python
graph = {
'A': {'B': 4, 'C': 2},
'B': {'A': 4, 'C': 5, 'D': 10},
'C': {'A': 2, 'B': 5, 'D': 3},
'D': {'B': 10, 'C': 3}
}

result = dijkstra(graph, 'A')

for node in result:
    print("Distance from A to", node, "=", result[node])
```

```python
def dijkstra_with_paths(graph, start):
    """
    Finds the shortest distances and paths from a source node to all other nodes
    in a weighted graph using Dijkstra's algorithm.
    Returns a tuple: (distances_dict, predecessors_dict).
    """
    distances = {node: float('inf') for node in graph}
    predecessors = {node: None for node in graph} # To store the path
    distances[start] = 0
    visited = set() # Use a set for faster lookup

    while len(visited) < len(graph):
        min_node = None
        min_distance = float('inf')

        # Find the unvisited node with the smallest distance
        for node in graph:
            if node not in visited and distances[node] < min_distance:
                min_distance = distances[node]
                min_node = node

        if min_node is None: # No reachable unvisited nodes left
            break

        visited.add(min_node)

        # Update distances and predecessors of neighbors
        for neighbor, weight in graph[min_node].items():
            new_distance = distances[min_node] + weight
            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                predecessors[neighbor] = min_node

    return distances, predecessors

def reconstruct_path(predecessors, start_node, end_node):
    """
    Reconstructs the shortest path from start_node to end_node using the
    predecessors dictionary.
    Returns a list of nodes forming the path, or None if no path exists.
    """
    path = []
    current = end_node
    while current is not None:
        path.insert(0, current)
        if current == start_node:
            break
        current = predecessors[current]

    if path and path[0] == start_node: # Path found and starts at source
        return path
    return None # No path found

def get_user_graph_input():
    """
    Prompts the user to input graph details and returns a dictionary representation.
    """
    graph = {}
    while True:
        num_nodes_str = input("Enter the number of nodes in the graph: ")
        if num_nodes_str.isdigit() and int(num_nodes_str) > 0:
            num_nodes = int(num_nodes_str)
            break
        else:
            print("Invalid input. Please enter a positive integer for the number of nodes.")

    nodes = []
    for i in range(num_nodes):
        while True:
            node_name = input(f"Enter unique name for node {i+1}: ").strip().upper()
            if node_name and node_name not in nodes:
                nodes.append(node_name)
                graph[node_name] = {}
                break
            elif not node_name:
                print("Node name cannot be empty. Please try again.")
            else:
                print(f"Node '{node_name}' already exists. Please enter a unique name.")

    print("\n--- Enter edges and weights for each node ---")
    print("Type 'done' for neighbor input when finished with a node's connections.\n")
    for node in nodes:
        print(f"For node '{node}':")
        while True:
            neighbor_input = input(f"  Enter neighbor of '{node}' (or 'done'): ").strip().upper()
            if neighbor_input == 'DONE':
                break
            if neighbor_input == node:
                print("A node cannot be a neighbor to itself in this context. Please enter a different node.")
                continue
            if neighbor_input not in nodes:
                print(f"Node '{neighbor_input}' does not exist in your graph. Please enter an existing node.")
                continue

            while True:
                weight_str = input(f"  Enter weight from '{node}' to '{neighbor_input}': ").strip()
                if weight_str.isdigit() and int(weight_str) >= 0:
                    weight = int(weight_str)
                    graph[node][neighbor_input] = weight
                    break
                else:
                    print("Invalid weight. Please enter a non-negative number.")
    return graph

# --- Main execution block ---
print("Welcome to Dijkstra's Algorithm implementation with user input!")

# 1. Get user input graph
user_graph = get_user_graph_input()
print("\nYour constructed graph:")
for node, neighbors in user_graph.items():
    print(f"  {node}: {neighbors}")

# 2. Test algorithm with different source nodes
while True:
    source_node = input("\nEnter the source node to start calculations from (or 'quit' to exit): ").strip().upper()
    if source_node == 'QUIT':
        break
    if source_node not in user_graph:
        print(f"Source node '{source_node}' not found in graph. Please try again.")
        continue

    distances, predecessors = dijkstra_with_paths(user_graph, source_node)

    print(f"\nShortest distances from source '{source_node}':")
    for node, dist in distances.items():
        if dist == float('inf'):
            print(f"  To {node}: Unreachable")
        else:
            print(f"  To {node}: {dist}")

    # Display shortest path with total distance for specific target nodes
    while True:
        target_node = input(f"\nEnter target node to show path from '{source_node}' (or 'done' to choose new source): ").strip().upper()
        if target_node == 'DONE':
            break
        if target_node not in user_graph:
            print(f"Target node '{target_node}' not found in graph. Please try again.")
            continue

        if distances[target_node] == float('inf'):
            print(f"No path from '{source_node}' to '{target_node}'.")
        else:
            path = reconstruct_path(predecessors, source_node, target_node)
            if path:
                print(f"Shortest path from '{source_node}' to '{target_node}': {' -> '.join(path)}")
                print(f"Total distance: {distances[target_node]}")
            else:
                print(f"Could not reconstruct path from '{source_node}' to '{target_node}'. (This should not happen if distance is finite)")

print("\nThank you for using the Dijkstra algorithm tool!")
```
