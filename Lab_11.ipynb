# Travelling Salesman Problem using Dynamic Programming.
Problem Description:
Given a set of cities and distances, find the shortest possible route that visits each city once and returns to the starting city.

#Theory
TSP uses dynamic programming with bit masking. Store results of visited subsets to reduce repeated work.

## Algorithm
1. Start from source city
2. Mark visited cities using bit mask
3. Try all unvisited cities
4. Choose minimum cost path
5. Return to source city

```python
def tsp(graph):
    n = len(graph)
    dp = [[-1]*(1<<n) for _ in range(n)]

    def visit(city, mask):
        if mask == (1<<n) - 1:
            return graph[city][0]

        if dp[city][mask] != -1:
            return dp[city][mask]

        ans = float('inf')

        for next_city in range(n):
            if mask & (1<<next_city) == 0:
                new_cost = graph[city][next_city] + visit(next_city, mask | (1<<next_city))
                if new_cost < ans:
                    ans = new_cost

        dp[city][mask] = ans
        return ans

    return visit(0, 1)
```

```python
graph = [
[0, 10, 15, 20],
[10, 0, 35, 25],
[15, 35, 0, 30],
[20, 25, 30, 0]
]

print("Minimum travelling cost:", tsp(graph))
```

```python
import sys
import heapq

def bellman_ford(graph, src):
    V = len(graph)
    dist = [sys.maxsize] * V
    parent = [-1] * V
    dist[src] = 0

    # Relax edges V-1 times
    for _ in range(V-1):
        for u in range(V):
            for v in range(V):
                if graph[u][v] != 0 and dist[u] != sys.maxsize and dist[u] + graph[u][v] < dist[v]:
                    dist[v] = dist[u] + graph[u][v]
                    parent[v] = u

    # Detect negative cycle
    for u in range(V):
        for v in range(V):
            if graph[u][v] != 0 and dist[u] != sys.maxsize and dist[u] + graph[u][v] < dist[v]:
                print("Graph contains a negative weight cycle!")
                return None, None

    return dist, parent

def print_path(parent, j):
    if parent[j] == -1:
        return [j]
    return print_path(parent, parent[j]) + [j]

def dijkstra(graph, src):
    V = len(graph)
    dist = [sys.maxsize] * V
    parent = [-1] * V
    dist[src] = 0
    pq = [(0, src)]
    while pq:
        d, u = heapq.heappop(pq)
        for v in range(V):
            if graph[u][v] != 0 and dist[u] + graph[u][v] < dist[v]:
                dist[v] = dist[u] + graph[u][v]
                parent[v] = u
                heapq.heappush(pq, (dist[v], v))
    return dist, parent

# ---- MAIN PROGRAM ----
n = int(input("Enter number of cities: "))
print("Enter adjacency matrix (use 0 if no edge):")
graph = []
for i in range(n):
    row = list(map(int, input().split()))
    graph.append(row)

src = int(input("Enter source city index (0-based): "))
dest = int(input("Enter destination city index (0-based): "))

# Bellman-Ford
dist, parent = bellman_ford(graph, src)
if dist:
    print("\nBellman-Ford:")
    print("Shortest distance:", dist[dest])
    path = print_path(parent, dest)
    print("Path of cities visited:", path)

    # Dijkstra
    dijkstra_dist, dijkstra_parent = dijkstra(graph, src)
    print("\nDijkstra:")
    print("Shortest distance:", dijkstra_dist[dest])
    path2 = print_path(dijkstra_parent, dest)
    print("Path of cities visited:", path2)
```
